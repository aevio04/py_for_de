```python
from datetime import datetime
moscow_str = 'Wednesday, October 2, 2002'
guardian_str = 'Friday, 11.10.13'
daily_news_str = 'Thursday, 18 August 1977'

moscow_dt = datetime.strptime(moscow_str, '%A, %B %d, %Y')
guardian_dt = datetime.strptime(guardian_str, '%A, %d.%m.%y')
daily_news_dt = datetime.strptime(daily_news_str, '%A, %d %B %Y')
print(moscow_dt, guardian_dt, daily_news_dt, sep = '\n')
```

```python
from datetime import datetime, timedelta
start_date = datetime.strptime('2022-01-01', '%Y-%m-%d')
end_date = datetime.strptime('2022-01-03', '%Y-%m-%d') 

date = start_date

while date <= end_date:
    print(date.strftime('%Y-%m-%d'))
    date += timedelta(days = 1)
```
